## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

### Results

The baseline produced a ranked action queue for all 30,000 rows in the dataset. The scoring rule prioritised content that was both stale (`days_since_last_update >= 180`) and visible in search (`impressions_90d >= 500`). The resulting queue was ranked using the baseline score, and the highest-ranked items were reviewed manually.

The baseline is useful as a transparent starting point because each recommendation can be explained using simple reason codes such as `stale_but_visible`, `not_stale`, and `low_visibility`.

A model-versus-baseline performance comparison is not reported here because the available baseline work does not contain a trained model evaluated on the same validation split. Therefore, no model performance numbers are claimed or invented.


## 5. Limitations

*What this work cannot claim.*

### Limitations

* The baseline uses only two features: `days_since_last_update` and `impressions_90d`.
* A high-priority item may still be a weak recommendation because high search visibility does not prove that the content needs a refresh.
* The ranked queue is decision-support, not an automatic decision. A person should review the recommendation before taking action.
* The available work does not establish that refreshing a page will cause better search performance.
* The leakage check confirmed that `trend_direction`, `trend_pct`, and `is_declining_label` were not used by the baseline.
* The results should therefore be interpreted as observed and directional rather than as proof of causal impact.


## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

### Ranked recommendations

1. **Review stale and visible content first** — prioritise items that are at least 180 days old and have at least 500 impressions in the last 90 days.
2. **Use the baseline score to create a review queue** — higher scores can be reviewed earlier because they combine staleness with search visibility.
3. **Check each recommendation manually** — confirm that the content actually needs an update before taking action.
4. **Use the reason code to understand the recommendation** — `stale_but_visible`, `not_stale`, and `low_visibility` provide a simple explanation for the baseline decision.
5. **Do not treat the ranking as proof of impact** — the baseline identifies candidates for review; it does not prove that refreshing content will improve search performance.
