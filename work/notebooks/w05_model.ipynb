### Method choice and why

I will use a **Decision Tree** model for this task. A Decision Tree is suitable because it is simple to understand and can capture non-linear relationships in the data. It also provides clear feature-based decision rules, which makes the model easier to interpret.

I will compare the Decision Tree with the Week-4 baseline using the **same data split and the same evaluation metric**. The goal is not to use a more complex model just for the sake of complexity, but to check whether the model improves on the existing baseline.


## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

```python
from sklearn.model_selection import train_test_split
```

```python
!git clone https://github.com/sriharshini-dev/flyrank-ml-internship-2026.git
```

```python
import pandas as pd

df = pd.read_csv(
    "/content/flyrank-ml-internship-2026/data/raw/content_refresh_anonymized.csv"
)

print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
from sklearn.model_selection import train_test_split

# Features used for the model
features = ["days_since_last_update", "impressions_90d"]

# Target based on the Week-4 baseline rule
target = (
    (df["days_since_last_update"] >= 180) &
    (df["impressions_90d"] >= 500)
).astype(int)

X = df[features].copy()
y = target.copy()

# 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Features:", features)
print("\nTarget distribution:")
print(y.value_counts())
```

## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Decision Tree Results")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, zero_division=0))
print("Recall:", recall_score(y_test, y_pred, zero_division=0))
print("F1 Score:", f1_score(y_test, y_pred, zero_division=0))
```

```python
# Week-4 baseline predictions on the same test set

baseline_pred = (
    (X_test["days_since_last_update"] >= 180) &
    (X_test["impressions_90d"] >= 500)
).astype(int)

print("Week-4 Baseline Results")
print("Accuracy:", accuracy_score(y_test, baseline_pred))
print("Precision:", precision_score(y_test, baseline_pred, zero_division=0))
print("Recall:", recall_score(y_test, baseline_pred, zero_division=0))
print("F1 Score:", f1_score(y_test, baseline_pred, zero_division=0))

print("\nComparison")
print("Decision Tree F1:", f1_score(y_test, y_pred, zero_division=0))
print("Baseline F1:", f1_score(y_test, baseline_pred, zero_division=0))
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

```python
# Check where the Decision Tree makes mistakes

comparison = pd.DataFrame({
    "actual": y_test,
    "predicted": y_pred
})

errors = comparison[comparison["actual"] != comparison["predicted"]]

print("Total test rows:", len(comparison))
print("Total errors:", len(errors))
print("Error rate:", len(errors) / len(comparison))
```

```python
# Show the Decision Tree feature importance

importance = pd.DataFrame({
    "feature": features,
    "importance": model.feature_importances_
})

importance = importance.sort_values(
    "importance",
    ascending=False
)

importance
```
