### Research Question

Which pages should be prioritized for human review based on available content and search-performance signals?

### Decision

The analysis supports the decision of which pages should be reviewed first for a possible content refresh or another appropriate action.

### Action

A content or SEO team member can review the prioritized pages and decide whether to refresh, expand, protect, prune, or monitor them.

### Cost of a Wrong Recommendation

A wrong recommendation could waste review time on pages that do not need attention. It could also lead to an unnecessary change to a page that is already performing well.

### Unit of Analysis

The unit of analysis is an individual content page.
