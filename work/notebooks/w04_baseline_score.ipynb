## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# Build the baseline score

df["stale"] = (df["days_since_last_update"] >= 180).astype(int)
df["visible"] = (df["impressions_90d"] >= 500).astype(int)

# Transparent score:
# Higher impressions among stale content gets higher priority.
df["score"] = df["stale"] * df["visible"] * df["impressions_90d"]

# Reason code
df["reason_code"] = "not_stale"

df.loc[
    (df["stale"] == 1) & (df["visible"] == 1),
    "reason_code"
] = "stale_but_visible"

df.loc[
    (df["stale"] == 1) & (df["visible"] == 0),
    "reason_code"
] = "low_visibility"

# Action
df["action"] = "monitor"

df.loc[
    df["reason_code"] == "stale_but_visible",
    "action"
] = "refresh"

# Rank the content
ranked = df.sort_values(
    ["score", "impressions_90d"],
    ascending=[False, False]
).copy()

ranked["rank"] = range(1, len(ranked) + 1)

# Keep the important columns
output = ranked[
    [
        "rank",
        "content_id",
        "score",
        "action",
        "reason_code",
        "days_since_last_update",
        "impressions_90d"
    ]
]

# Write the required CSV
output.to_csv(
    "/content/flyrank-ml-internship-2026/work/outputs/baseline_action_score.csv",
    index=False
)

print("CSV created successfully.")
print("Rows:", len(output))
print("\nTop 10:")
print(output.head(10))
```
